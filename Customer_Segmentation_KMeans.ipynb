```python
import pandas as pd

file_path = '/content/drive/My Drive/Colab Notebooks/segmentation.csv'# Ganti dengan jalur sebenarnya

data = pd.read_csv(file_path)

data.head()
```

```python
print(data.info)

print(data.describe)
```

```python
#mengapus data yang dianggap tidak berguna
data_cleaned = data.drop(columns=['ID','Education','Settlement size'])


#menampilkan datset
print("\nDtaaset setelah menghapus atribut 'ID,education,dan settlement size'")
print(data_cleaned.head())
```

```python
from sklearn.cluster import KMeans

model = KMeans(n_clusters=4,random_state=42)

model.fit(data_cleaned)

data_cleaned['Cluster'] = model.labels_
```

```python
data_cleaned['Cluster'].value_counts()
```

```python
from sklearn.metrics import silhouette_score

#mengenerate silhoutte score
silhouette_avg = silhouette_score(data_cleaned, data_cleaned['Cluster'])
print(f"Silhouette Score: {silhouette_avg}") # untuk melihat cluster sudah benar apa belum
#apabila nilainya mendekati satu disebut sebagai yang terbaik
```

```python
# prompt: kode matriks evaluasi davies-bouldin index

from sklearn.metrics import davies_bouldin_score

davies_bouldin = davies_bouldin_score(data_cleaned, data_cleaned['Cluster'])
print(f"Davies-Bouldin Index: {davies_bouldin}")# semakin kecil semakin baik
```

```python
scores_matrix = []

# Looping untuk jumlah cluster
for n_clusters in range(2, 11):  # Misal, mencoba dari 2 hingga 10 clusters
  kmeans = KMeans(n_clusters=n_clusters, random_state=42, n_init=10)
  kmeans.fit(data_cleaned)
  labels = kmeans.labels_

  # Menghitung skor
  silhouette_avg = silhouette_score(data_cleaned, labels)
  davies_bouldin_index = davies_bouldin_score(data_cleaned, labels)

  scores_matrix.append([n_clusters, silhouette_avg, davies_bouldin_index])


# Membuat dataframe untuk memudahkan visualisasi
scores_df = pd.DataFrame(scores_matrix, columns=["Number of Clusters", "Silhouette Score", "Davies-Bouldin Index"])
scores_df
```
